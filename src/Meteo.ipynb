```python
import pandas as pd

# Load the data from the text file
file_path = '/content/2025-09-04ECS.txt'
# file_path = '/content/test.csv'

# Read the file line by line and extract data
data = []
with open(file_path, 'r') as f:
    # Skip the header and units lines
    next(f)
    next(f)
    for line in f:
        parts = line.split(',')
        # Check if the line has enough columns before accessing index 5
        if len(parts) > 5:
            # Extract Date, Time, and NO2. Extract only the time part from the Time column. Convert NO2 to float.
            data.append({
                'Date': parts[0].split()[0],
                'Time': parts[0].split()[1], # Extract only the time part
                'NO2': float(parts[1])  # NO2 is the 6th column (index 5)
            })

# Create a pandas DataFrame
df = pd.DataFrame(data)

# Display the first few rows of the DataFrame
display(df.head())
```

```python
# Combine Date and Time into a single datetime column
df['DateTime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'])

# Display the first few rows with the new column
display(df.head())
```

```python
import matplotlib.pyplot as plt

# Create the time series plot
plt.figure(figsize=(12, 6))
plt.plot(df['DateTime'], df['NO2'])
plt.xlabel('Date and Time')
plt.ylabel('NO2 Level')
plt.title('NO2 Level Over Time')
plt.grid(True)
plt.show()
```
